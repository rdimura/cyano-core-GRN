```python
import pandas as pd
import networkx as nx
from sklearn.preprocessing import MinMaxScaler
```

```python
xlsx=pd.ExcelFile("grn_1100_min9_annotation_03_25_24.xlsx")
df=pd.read_excel(xlsx,'Transcription regulators')
#making the integrated value
df['BC_normalized']=df['BetweennessCentrality']/df['BetweennessCentrality'].max()
df['CC_normalized']=df['ClosenessCentrality']/df['ClosenessCentrality'].max()
df['Deg_normalized']=df['degree']/df['degree'].max()
df['EVC_normalized']=df['eigenvector centrality']/df['eigenvector centrality'].max()
df['KC_normalized']=df['k-core']/df['k-core'].max()
df['ST_normalized']=df['Stress']/df['Stress'].max()
df['integrated_value']=df['BC_normalized']+df['CC_normalized']+df['Deg_normalized']+df['EVC_normalized']+df['KC_normalized']+df['KC_normalized']+df['ST_normalized']

```

```python
#automates many columns that are useless to me being dropped
#7942 has an excess of columns that at this point its better to do it by hand
df.drop(['BC_normalized','CC_normalized','Deg_normalized','EVC_normalized','KC_normalized','ST_normalized'], axis=1, inplace=True)
df.drop(['class','seq_type','chromosome','genomic_accession','start','end','strand','product_accession','GeneID','feature_interval_length'],axis=1,inplace=True)
df.drop(['product_length','AverageShortestPathLength','ClusteringCoefficient','Eccentricity','EdgeCount','Indegree','Outdegree','IsSingleNode','NeighborhoodConnectivity'],axis=1, inplace=True)
df.drop(['PartnerOfMultiEdgedNodePairs','selected','SelfLoops','shared name'], axis=1,inplace=True)
```

```python
df.to_csv('core_grn_min9_w_IV_03_25_24.csv')
```
